```python
import nltk
from nltk.corpus import wordnet as wn
```

```python
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
#implementation of the lesk alg
def lesk(word, sentence):
  best_sense=None
  max_overlap=0
  sentence_tokens = set(nltk.word_tokenize(sentence))
  #get all senses of the word
  for sense in wn.synsets(word):
    #get the definition(gloss) and examples of the sense
    definition=set(nltk.word_tokenize(sense.definition()))
    examples= set(nltk.word_tokenize(' '.join(sense.examples())))

    #calculate overlap with the sentence
    overlap = len(definition.intersection(sentence_tokens)) + len(examples.intersection(sentence_tokens))
    if overlap > max_overlap:
      max_overlap = overlap
      best_sense = sense

  return best_sense

```

```python
sentence = "i went to the river bank to swim."
ambiguous_word = "swim"

sense = lesk(ambiguous_word, sentence)
print(f"the best sense for '{ambiguous_word}' in the given context is: {sense.name()}-{sense.definition()}")
```

```python
dataset = [
    ("I went to the bank to deposit the money.", "bank", "bank.n.01"),
    ("He has a bank of knowledge.", "bank", "bank.n.02"),
]

correct_prediction = 0
for sentence_text, ambiguous_word, correct_sense_name in dataset:
  predicted_sense = lesk(ambiguous_word, sentence_text)
  if predicted_sense and predicted_sense.name() == correct_sense_name:
    correct_prediction+=1

accuracy = correct_prediction/len(dataset)
print(f"accuracy:{accuracy:.2f}")
```
