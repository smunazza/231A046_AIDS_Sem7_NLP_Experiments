```python
documents = ["machine learning enables computers to learn from data.",
             "computers can learn patterns from data using machine learning.",
             "the football team won the match yesterday.",
             "deep learning is a branch of machine learning.",
             "natural language processing helps computers understand human language."]
```

```python
 import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
for i, document in enumerate(documents):
  print(f"D{i+1}: {document}")
```

```python
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(documents)
print(tfidf_matrix.shape)
```

```python
print(vectorizer.get_feature_names_out())
```

```python
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=["D1","D2","D3","D4","D5"]
)
tfidf_df.round(3)
```

```python
similarity_matrix = cosine_similarity(tfidf_matrix)
similarity_matrix
```

```python
similarity_df=pd.DataFrame(
    similarity_matrix,
    columns=["D1","D2","D3","D4","D5"],
    index=["D1","D2","D3","D4","D5"]
)
similarity_df.round(3)
```

```python
d1=0
d2=1
d3=2
d4=3
```

```python
doc2=1
doc4=3

similarity=similarity_matrix[doc2][doc4]
print(f"similarity between D{doc2+1} and D{doc4+1}: "
f"{similarity:.3f}")
```

```python
highest_similarity=-1
most_similar_pair=None

for i in range(len(documents)):
  for j in range(i+1, len(documents)):
    similarity=similarity_matrix[i][j]
    if similarity>highest_similarity:
      highest_similarity=similarity
      most_similar_pair=(i,j)

print(
    f"most similar documents: "
    f" D{most_similar_pair[0]+1} and D{most_similar_pair[1]+1}"
)
print(
    f"similarity: {highest_similarity:.3f}"
)
```

```python
def get_document_similarity(doc_index_1, doc_index_2):
  if not (0 <= doc_index_1 < len(documents) and 0 <= doc_index_2 < len(documents)):
    raise IndexError("Document indices are out of bounds.")

  return similarity_matrix[doc_index_1][doc_index_2]

sim_d1_d2 = get_document_similarity(0, 1)
print(f"Similarity between Document 1 and Document 2: {sim_d1_d2:.3f}")

sim_d4_d5 = get_document_similarity(3, 4)
print(f"Similarity between Document 4 and Document 5: {sim_d4_d5:.3f}")
```
