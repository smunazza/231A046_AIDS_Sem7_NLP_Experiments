```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag, UnigramTagger, BigramTagger
from nltk.corpus import treebank
from nltk import pos_tag_sents

```

```python
nltk.download('treebank')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt_tab')
```

```python
sample_text="the quick brown fox jumps over the lazy dog. the lazy dog kept sleeping."

tokens = word_tokenize(sample_text)
```

```python
rule_based_tags=pos_tag(tokens)
```

```python
print(rule_based_tags)
```

```python
nltk.download('tagsets_json')
nltk.help.upenn_tagset('VBZ')
```

```python
train_sents=treebank.tagged_sents()[:3000]
unigram_tagger=UnigramTagger(train_sents)
unigram_tags=unigram_tagger.tag(tokens)

bigram_tagger=BigramTagger(train_sents, backoff=unigram_tagger)
bigram_tags=bigram_tagger.tag(tokens)
```

```python
bigram_tags
```

```python
unigram_tags
```

```python
print("original sentence:, sample_text")
print("\nrule-based pos tagging(statistical):")
print(rule_based_tags)
print("\nunigram pos tagging:")
print(unigram_tags)
print("\nbigram pos tagging:")
print(bigram_tags)
```

accuracy = no of correctly tagged words/ total no of words

```python
test_sents=treebank.tagged_sents()[3000:]
unigram_accuracy=unigram_tagger.accuracy(test_sents)
bigram_accuracy=bigram_tagger.accuracy(test_sents)
print(f"\\unigram accuracy:, {unigram_accuracy*100:.2f}%")
print(f"\\bigram accuracy:, {bigram_accuracy*100:.2f}%")
```
